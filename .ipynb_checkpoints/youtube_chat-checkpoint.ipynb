```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled

from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

```python
os.environ["OPENAI_API_KEY"] = "sk-xxxxxxxxxxxxxxxx"
```

```python
video_id = "B7Vz_wNSqr0"

try:
    transcript_list = YouTubeTranscriptApi().fetch(
        video_id,
        languages=["en"]
    )

    # Convert FetchedTranscript to string
    transcript = " ".join(chunk.text for chunk in transcript_list)

    print("Transcript type:", type(transcript))
    print("\nTranscript:\n")
    print(transcript)

except TranscriptsDisabled:
    print("Transcript is disabled for this video.")
    exit()

except Exception as e:
    print("Error while fetching transcript:", e)
    exit()
```

```python
# step -1 (text splitter)
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.create_documents([transcript])

print("\nNumber of chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0].page_content)
```

```python
# step-2 (embedding generation and storing in vector store)
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
)
```

```python
# step-3-Create FAISS Vector Store
```

```python
vector_store = FAISS.from_documents(
    chunks,
    embeddings
)

print("\nVector store created successfully.")
```

```python
# step -4-Create Retriever
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 10}
)
```

```python
#   Test Retriever
```

```python
test_question = "What is self introduction?"
retrieved_docs = retriever.invoke(test_question)
```

```python
print("\nRetrieved documents:\n")

for i, doc in enumerate(retrieved_docs):
    print(f"\n--- Document {i + 1} ---")
    print(doc.page_content)
```

```python
#  LLM
```

```python
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.2
)
```

```python
# Prompt
```

```python
prompt = PromptTemplate(
    template="""
You are an interview analysis assistant.

Answer ONLY using the provided transcript context.

Your task:
1. Identify all interview questions asked to the student.
2. Give the total number of questions.
3. List each question.
4. Give the student's answer for each question.
5. Check whether the student's answer is correct.
6. If the student's answer is wrong or incomplete, provide a corrected answer.
7. Do not invent questions or answers that are not present in the transcript.

If the provided context does not contain enough information, say:
"I don't know based on the provided transcript."

Context:
{context}

Question:
{question}
""",
    input_variables=["context", "question"]
)
```

```python
#  User Question
```

```python
question = """
Tell me how many questions were asked to the student.

Give me a list of all questions with:
- Question
- Student's answer
- Whether the answer is correct
- Correct answer if the student's answer is wrong or incomplete
"""
```

```python
# Retrieve relevant documents
```

```python
retriever_docs = retriever.invoke(question)
```

```python
#  Create context
```

```python
context_text = "\n\n".join(
    doc.page_content
    for doc in retriever_docs
)
```

```python
# Create final prompt
```

```python
final_prompt = prompt.format(
    context=context_text,
    question=question
)
```

```python
#  Get final answer
```

```python
answer = llm.invoke(final_prompt)

print("\n\n================ FINAL ANSWER ================\n")
print(answer.content)
```
